# Latin SQL Banking Scenario

This notebook preserves standard SQLite syntax while using reviewed Latin names for tables, columns, and domain values. It evaluates one proposed banking exchange through an ATLED evidence record.

```python
import sqlite3
from pathlib import Path

scriptum = Path('../scenarios/banking_latine.sql').read_text(encoding='utf-8')
connexio = sqlite3.connect(':memory:')
connexio.executescript(scriptum)
```

```python
quaestio = '''
SELECT c.commutatio_id, o.nomen AS entitas_originis,
       d.nomen AS entitas_destinationis, c.quantitas,
       a.altruismus, a.tempus, a.longaevitas, a.energia,
       a.decisio, a.fiducia
FROM commutationes AS c
JOIN rationes AS ro ON ro.ratio_id = c.ratio_origo_id
JOIN entitates AS o ON o.entitas_id = ro.entitas_id
JOIN rationes AS rd ON rd.ratio_id = c.ratio_destinatio_id
JOIN entitates AS d ON d.entitas_id = rd.entitas_id
JOIN aestimationes_atled AS a ON a.commutatio_id = c.commutatio_id
'''

columnae = [item[0] for item in connexio.execute(quaestio).description]
eventus = [dict(zip(columnae, row)) for row in connexio.execute(quaestio)]
eventus
```

## Integrity rule

The translator may propose Latin terminology, but only reviewed glossary entries become schema identifiers. SQL keywords such as `SELECT`, `JOIN`, `CHECK`, and `CREATE TABLE` remain unchanged.
